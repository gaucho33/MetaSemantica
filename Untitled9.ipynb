```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from transformers import AutoTokenizer, AutoModelForCausalLM

class NeighborhoodHookSteerer:
    def __init__(
        self,
        model_name: str,
        target_layer_idx: int = 16,
        gamma_max: float = 6.0,
        h_target: float = 4.0,
        tau: float = 0.5
    ):
        self.tokenizer = AutoTokenizer.from_pretrained(model_name)
        self.model = AutoModelForCausalLM.from_pretrained(
            model_name,
            torch_dtype=torch.float16,
            device_map="auto"
        )
        self.model.eval()

        self.target_layer_idx = target_layer_idx
        self.gamma_max = gamma_max
        self.h_target = h_target
        self.tau = tau

        # Estadísticos de calibración y trayectoria
        self.mu_channel = None
        self.std_channel = None
        self.target_energy = None
        self.current_R_energy = None

        # Historial de telemetría para métricas
        self.telemetry_history = []

        # Handles para remover hooks al finalizar
        self.hooks = []
        self._attach_hooks()

    def _extract_residual_energy(self, hidden_state: torch.Tensor) -> torch.Tensor:
        """Calcula E_raw a partir de A inyectado por el hook [Batch, N, D]."""
        A = hidden_state.squeeze(0).float() # [N, D]
        A_pos = A - A.min() + 1e-8

        F_vec = A_pos.sum(dim=-1, keepdim=True)
        C_vec = A_pos.sum(dim=-2, keepdim=True)
        T_sum = A_pos.sum()
        M = torch.matmul(F_vec, C_vec) / T_sum
        R = A_pos - M

        return (R ** 2).mean(dim=0).unsqueeze(0) # [1, D]

    def _layer_hook(self, module, input, output):
        """Hook Extractor: Captura A en la capa intermedia deseada."""
        # En Transformers, output puede ser una tupla (hidden_states, ...)
        hidden_states = output[0] if isinstance(output, tuple) else output
        # Solo procesamos el último token en paso autorregresivo o la secuencia completa
        self.current_R_energy = self._extract_residual_energy(hidden_states)

    def _lm_head_hook_optimized(self, module, input, output):
      if self.target_energy is None or self.current_R_energy is None:
          return output

      logits = output
    # 1. Entropía pura en Tensor GPU (sin .item() en el hot path)
      probs = F.softmax(logits[:, -1, :], dim=-1)
      log_probs = F.log_softmax(logits[:, -1, :], dim=-1)
      H_t = -torch.sum(probs * log_probs, dim=-1) / 0.693147  # ln(2) precalculado

    # 2. Gamma sigmoidal sin salir de CUDA
      h_diff = (H_t - self.h_target) / self.tau
      gamma_t = self.gamma_max * torch.sigmoid(h_diff).unsqueeze(-1)  # [1, 1]

    # 3. Proyección sobre vocabulario en precisión nativa
      W_head = module.weight
      Delta_E = F.normalize(self.target_energy, p=2, dim=-1).to(W_head.dtype)
      vocab_bias = torch.matmul(Delta_E, W_head.T)

    # 4. Modulación directa sin casting superfluo
      logits[:, -1, :] = logits[:, -1, :] + (gamma_t * vocab_bias)

      # Almacenar métricas en el historial de telemetría como tensores
      self.telemetry_history.append({"H_t": H_t, "gamma_t": gamma_t})

      return logits

    def _attach_hooks(self):
        """Registra los hooks en la arquitectura del modelo."""
        # Detectar la lista de capas según la arquitectura (Llama, Qwen, GPT-NeoX)
        if hasattr(self.model, "model"):
            layers = self.model.model.layers
        elif hasattr(self.model, "transformer"):
            layers = self.model.transformer.h
        else:
            raise AttributeError("Arquitectura no reconocida para enganchar capas ocultas.")

        # Hook 1: Extractor en Capa Intermedia
        h1 = layers[self.target_layer_idx].register_forward_hook(self._layer_hook)
        # Hook 2: Modulador en la lm_head
        h2 = self.model.get_output_embeddings().register_forward_hook(self._lm_head_hook_optimized)

        self.hooks.extend([h1, h2])

    def calibrate(self, neutral_corpus: list[str]):
        energies = []
        with torch.no_grad():
            for text in neutral_corpus:
                inputs = self.tokenizer(text, return_tensors="pt").to(self.model.device)
                self.model(**inputs)
                energies.append(self.current_R_energy)

        stack = torch.cat(energies, dim=0)
        self.mu_channel = stack.mean(dim=0, keepdim=True)
        self.std_channel = stack.std(dim=0, keepdim=True) + 1e-8

    def set_target(self, anchor_texts: list[str]):
        whitened_vectors = []
        with torch.no_grad():
            for text in anchor_texts:
                inputs = self.tokenizer(text, return_tensors="pt").to(self.model.device)
                self.model(**inputs)
                E_w = (self.current_R_energy - self.mu_channel) / self.std_channel
                whitened_vectors.append(E_w)

        stack = torch.cat(whitened_vectors, dim=0)
        self.target_energy = F.normalize(stack.mean(dim=0, keepdim=True), p=2, dim=1)

    def disable_steering(self):
        """Desactiva temporalmente el steering seteando target_energy a None."""
        self.target_energy = None

    def remove_hooks(self):
        """Limpia los hooks instalados para restaurar el modelo nativo."""
        for hook in self.hooks:
            hook.remove()
        self.hooks.clear()
```

```python
import time
import torch
import pandas as pd
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch.nn.functional as F

# =====================================================================
# SCRIPT DE PRUEBA Y METRICAS COMPARATIVAS (A/B TEST)
# =====================================================================
def run_ab_test(
    model_id: str = "Qwen/Qwen2.5-1.5B-Instruct",
    target_layer_idx: int = 14,
    gamma_max: float = 7.0,
    neutral_corpus: list[str] = None,
    anchor_texts: list[str] = None,
    prompt: str = "El comportamiento de la materia a altas temperaturas demuestra que",
    max_tokens: int = 45,
    temperature: float = 0.7
):
    if neutral_corpus is None:
        neutral_corpus = [
            "El gato duerme sobre la silla del jardín.",
            "La manzana cayó del árbol en una tarde tranquila.",
            "El agua transparente fluye por el arroyo de la montaña.",
            "La casa de ladrillos tiene tres ventanas cuadradas.",
            "El autobús público llega a la estación central a horario."
        ]
    if anchor_texts is None:
        anchor_texts = [
            "La simetría gauge del espacio de hilbert impone restricciones fermiónicas.",
            "El tensor de curvatura de ricci cuantifica el gradiente de energía latente.",
            "La descomposición espectral refleja el invariante topológico en el manifold."
        ]

    print(f"Cargando modelo {model_id} con hooks inyectados...")
    # La clase NeighborhoodHookSteerer se importa o define en una celda anterior.
    steerer = NeighborhoodHookSteerer(model_name=model_id, target_layer_idx=target_layer_idx, gamma_max=gamma_max)

    # 1. Corpus Neutro (Calibración)
    steerer.calibrate(neutral_corpus)

    # 2. Firma Objetivo (Cosema Técnico / Formulación Física Estructurada)
    # steerer.set_target(anchor_texts) # This line was moved below for logical flow.

    # --- CORRIDA A: NATIVA (Sin Target Inyectado) ---
    steerer.disable_steering() # Ensure steering is disabled for the native run
    inputs = steerer.tokenizer(prompt, return_tensors="pt").to(steerer.model.device)

    t0 = time.perf_counter()
    with torch.no_grad():
        out_native = steerer.model.generate(
            **inputs,
            max_new_tokens=max_tokens,
            do_sample=True,
            temperature=temperature
        )
    t_native = (time.perf_counter() - t0) * 1000.0
    text_native = steerer.tokenizer.decode(out_native[0], skip_special_tokens=True)

    # --- CORRIDA B: DIRECCIONADA (Hook Steering Activo) ---
    steerer.set_target(anchor_texts) # Set target here, before steered generation
    steerer.telemetry_history.clear()

    t0 = time.perf_counter()
    with torch.no_grad():
        out_steered = steerer.model.generate(
            **inputs,
            max_new_tokens=max_tokens,
            do_sample=True,
            temperature=temperature
        )
    t_steered = (time.perf_counter() - t0) * 1000.0
    text_steered = steerer.tokenizer.decode(out_steered[0], skip_special_tokens=True)

    # Extraer métricas de telemetría del Steerer, moviendo tensores a CPU y obteniendo el valor
    history = steerer.telemetry_history if hasattr(steerer, 'telemetry_history') else []
    avg_h = sum(item["H_t"].cpu().item() for item in history) / len(history) if history else 0.0
    avg_g = sum(item["gamma_t"].cpu().item() for item in history) / len(history) if history else 0.0

    # Imprimir Tabla de Resultados
    metrics = [
        {
            "Modo": "Nativo (Sin Steerer)",
            "Tiempo Total (ms)": round(t_native, 2),
            "Latencia (ms/token)": round(t_native / max_tokens, 2),
            "γ Promedio": 0.0,
            "Entropía Media (bits)": "N/A"
        },
        {
            "Modo": "Direccionado (Con Hook)",
            "Tiempo Total (ms)": round(t_steered, 2),
            "Latencia (ms/token)": round(t_steered / max_tokens, 2),
            "γ Promedio": round(avg_g, 2),
            "Entropía Media (bits)": round(avg_h, 2)
        }
    ]

    print("\n" + "="*80)
    print("COMPARATIVA DE GENERACIÓN EN VIVO (A/B TEST)")
    print("="*80)
    print(f"\n[PROMPT INICIAL]: '{prompt}'\n")
    print(f"--- GENERACIÓN NATIVA ---\n{text_native}\n")
    print(f"--- GENERACIÓN DIRECCIONADA (HOOK STEERED) ---\n{text_steered}\n")
    print("="*80)
    print(pd.DataFrame(metrics).to_string(index=False))
```

## Empaquetado para Producción en Masa: `ProductionHookSteerer`

La clase `ProductionHookSteerer` está optimizada para el despliegue en entornos de producción, priorizando la eficiencia y la mínima latencia. Aquí se detalla cómo está estructurada y cómo se integraría en un flujo de trabajo típico de inferencia de modelos de lenguaje.

### Características Clave para Producción:

1.  **Núcleo Compilado con `torch.compile`**: La lógica central de steering (`_compiled_steering_kernel`) está diseñada para ser compilada por PyTorch (si el entorno lo permite), fusionando operaciones en un único kernel de GPU para maximizar el rendimiento.
2.  **Cero `cpu().item()` en el Hot Path**: Evita la costosa sincronización entre CPU y GPU durante la generación, lo que es crítico para la latencia.
3.  **Pre-cálculo de Bias de Vocabulario**: El término `vocab_bias` (la proyección del `target_energy` sobre el vocabulario del modelo) se calcula una única vez cuando se establece el objetivo (`set_target`), y se almacena en memoria de GPU. Esto elimina cálculos repetitivos en cada paso de generación.
4.  **Activación Controlada**: Permite activar y desactivar el steering de forma programática (`enable()` y `disable()`) sin necesidad de desenganchar y reenganchar los hooks, lo que es útil para escenarios A/B testing en producción o para alternar comportamientos del modelo.

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from transformers import AutoTokenizer, AutoModelForCausalLM

# =====================================================================
# KERNEL COMPILADO (JIT / PyTorch 2.0+)
# =====================================================================
# Se recomienda el uso de torch.compile para PyTorch 2.0+ para optimizar al máximo.
# Si no se usa PyTorch 2.0+, la función se ejecutará normalmente sin compilación.
@torch.compile(mode="max-autotune", fullgraph=True)
def _compiled_steering_kernel(
    logits: torch.Tensor,
    vocab_bias: torch.Tensor,
    gamma_max: float,
    h_target: float,
    tau: float
) -> torch.Tensor:
    """
    Kernel compilado en grafo cerrado. Fusiona el cálculo de entropía de Shannon,
    la modulación sigmoidal y la inyección tensorial sobre los logits en un solo paso GPU.
    """
    last_logits = logits[:, -1, :]

    # Entropía de Shannon base 2 (0.6931471805599453 = ln(2))
    probs = F.softmax(last_logits, dim=-1)
    log_probs = F.log_softmax(last_logits, dim=-1)
    H_t = -torch.sum(probs * log_probs, dim=-1) / 0.6931471805599453

    # Control sigmoidal en tensor puro
    h_diff = (H_t - h_target) / tau
    gamma_t = gamma_max * torch.sigmoid(h_diff).unsqueeze(-1)

    # Inyección in-place/fusionada sobre los logits del último token
    steered_logits = last_logits + (gamma_t * vocab_bias)
    logits_out = logits.clone()
    logits_out[:, -1, :] = steered_logits
    return logits_out


# =====================================================================
# CLASE PRINCIPAL DE PRODUCCIÓN: ProductionHookSteerer
# =====================================================================
class ProductionHookSteerer:
    """
    Middleware de steering para producción en PyTorch.
    - Cero sincronización CPU <-> GPU (.item() eliminados).
    - Precalculo de bias probabilístico fuera del loop autorregresivo.
    - Kernel de modulación fusionado vía torch.compile.
    """
    def __init__(
        self,
        model: nn.Module,
        tokenizer: AutoTokenizer,
        target_layer_idx: int = 14,
        gamma_max: float = 7.0,
        h_target: float = 3.8,
        tau: float = 0.5
    ):
        self.model = model
        self.tokenizer = tokenizer
        self.target_layer_idx = target_layer_idx
        self.gamma_max = gamma_max
        self.h_target = h_target
        self.tau = tau

        self.mu_channel = None
        self.std_channel = None
        self.target_energy = None

        # Tensor de bias precalculado en GPU: [1, Vocab]
        self.precomputed_vocab_bias = None
        self.is_active = False

        self.hooks = []
        self._attach_hooks()

    def _extract_residual_energy(self, hidden_state: torch.Tensor) -> torch.Tensor:
        """Descomposición R = A - M optimizada en tensores CUDA."""
        A = hidden_state.squeeze(0).to(torch.float32)
        A_pos = A - A.min() + 1e-8

        F_vec = A_pos.sum(dim=-1, keepdim=True)
        C_vec = A_pos.sum(dim=-2, keepdim=True)
        T_sum = A_pos.sum()

        M = torch.matmul(F_vec, C_vec) / T_sum
        R = A_pos - M
        return (R ** 2).mean(dim=0).unsqueeze(0)

    def _layer_hook(self, module, input, output):
        """Hook Extractor de baja sobrecarga en capa intermedia."""
        hidden_states = output[0] if isinstance(output, tuple) else output
        self.current_R_energy = self._extract_residual_energy(hidden_states)

    def _lm_head_hook(self, module, input, output):
        """Hook Modulador: Invoca el kernel compilado si el steering está activo."""
        if not self.is_active or self.precomputed_vocab_bias is None:
            return output

        return _compiled_steering_kernel(
            logits=output,
            vocab_bias=self.precomputed_vocab_bias,
            gamma_max=self.gamma_max,
            h_target=self.h_target,
            tau=self.tau
        )

    def _attach_hooks(self):
        """Registra los forward hooks en el modelo HuggingFace/PyTorch."""
        if hasattr(self.model, "model"):
            layers = self.model.model.layers
        elif hasattr(self.model, "transformer"):
            layers = self.model.transformer.h
        else:
            raise AttributeError("Arquitectura del modelo no soportada para hooks de capa.")

        h1 = layers[self.target_layer_idx].register_forward_hook(self._layer_hook)
        h2 = self.model.get_output_embeddings().register_forward_hook(self._lm_head_hook)
        self.hooks.extend([h1, h2])

    def calibrate(self, neutral_corpus: list[str]):
        """Calcula medias y desviaciones estándar del corpus neutro."""
        energies = []
        device = next(self.model.parameters()).device

        with torch.no_grad():
            for text in neutral_corpus:
                inputs = self.tokenizer(text, return_tensors="pt").to(device)
                self.model(**inputs)
                energies.append(self.current_R_energy)

        stack = torch.cat(energies, dim=0)
        self.mu_channel = stack.mean(dim=0, keepdim=True)
        self.std_channel = stack.std(dim=0, keepdim=True) + 1e-8

    def set_target(self, anchor_texts: list[str]):
        """
        Calcula la firma de energía e incrusta la proyección tensorial precalculada
        (Delta_E @ W_head.T) directamente en memoria GPU.
        """
        if self.mu_channel is None:
            raise ValueError("Se debe ejecutar calibrate() antes de fijar el target.")

        device = next(self.model.parameters()).device
        whitened_vectors = []

        with torch.no_grad():
            for text in anchor_texts:
                inputs = self.tokenizer(text, return_tensors="pt").to(device)
                self.model(**inputs)
                E_w = (self.current_R_energy - self.mu_channel) / self.std_channel
                whitened_vectors.append(E_w)

        stack = torch.cat(whitened_vectors, dim=0)
        self.target_energy = F.normalize(stack.mean(dim=0, keepdim=True), p=2, dim=1)

        # Precalculo del sesgo de vocabulario fuera del loop de generación
        lm_head = self.model.get_output_embeddings()
        W_head = lm_head.weight.detach()  # [Vocab, D]
        Delta_E = self.target_energy.to(dtype=W_head.dtype, device=W_head.device)

        # [1, D] x [D, Vocab] -> [1, Vocab]
        self.precomputed_vocab_bias = torch.matmul(Delta_E, W_head.T)
        self.is_active = True

    def disable(self):
        """Desactiva temporalmente el control sin desenganchar los hooks."""
        self.is_active = False

    def enable(self):
        """Reorganiza la activación del control."""
        if self.precomputed_vocab_bias is not None:
            self.is_active = True

    def remove_hooks(self):
        """Elimina definitivamente los hooks de PyTorch."""
        for h in self.hooks:
            h.remove()
        self.hooks.clear()
```

### Ejemplo de Uso de `ProductionHookSteerer`

Aquí se muestra cómo instanciar, calibrar y usar la `ProductionHookSteerer` para la generación de texto. Este es el patrón que usarías en un servicio de inferencia.

```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
import time

# 1. Cargar el modelo y tokenizer
model_name = "Qwen/Qwen2.5-1.5B-Instruct" # Puedes cambiar este modelo
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype=torch.float16, # Importante para eficiencia en GPU
    device_map="auto" # Distribuye el modelo automáticamente en las GPUs disponibles
)
model.eval()

# 2. Instanciar el ProductionHookSteerer
# Asegúrate de que los parámetros (target_layer_idx, gamma_max, etc.) sean adecuados para tu modelo y caso de uso.
steerer = ProductionHookSteerer(
    model=model,
    tokenizer=tokenizer,
    target_layer_idx=14, # Ajusta según el modelo específico
    gamma_max=7.0,
    h_target=3.8,
    tau=0.5
)

# 3. Definir corpus neutro y de anclaje (estos deben ser representativos de tus necesidades)
neutral_corpus_prod = [
    "El perro ladra al cartero todos los días.",
    "La lluvia cae suavemente sobre el tejado.",
    "El libro está abierto en la página cincuenta."
]
anchor_texts_prod = [
    "La superconductividad se observa en materiales a temperaturas criogénicas.",
    "La mecánica cuántica describe el comportamiento de la materia a escala atómica y subatómica.",
    "La relatividad general de Einstein reformuló la gravedad como una curvatura del espacio-tiempo."
]

# 4. Calibrar el steerer con el corpus neutro
print("Calibrando steerer...")
steerer.calibrate(neutral_corpus_prod)
print("Steerer calibrado.")

# 5. Establecer el objetivo de steering con los textos ancla
print("Estableciendo objetivo de steering...")
steerer.set_target(anchor_texts_prod)
print("Objetivo establecido. Steerer activo.")

# 6. Generar texto con el steerer activado
prompt = "El bosón de Higgs es una partícula fundamental en el modelo estándar que"
inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

print(f"\nGenerando texto con steering para el prompt: '{prompt}'")
t0 = time.perf_counter()
with torch.no_grad():
    output_steered = model.generate(
        **inputs,
        max_new_tokens=50,
        do_sample=True,
        temperature=0.7
    )
t_steered = (time.perf_counter() - t0) * 1000.0
text_steered = tokenizer.decode(output_steered[0], skip_special_tokens=True)
print(f"--- GENERACIÓN DIRECCIONADA ---\n{text_steered}")
print(f"Tiempo de generación (steered): {t_steered:.2f} ms\n")

# 7. Desactivar el steerer y generar texto nativo (opcional, para comparación)
steerer.disable()
print("Generando texto sin steering (desactivado) para el mismo prompt:")
t0 = time.perf_counter()
with torch.no_grad():
    output_native = model.generate(
        **inputs,
        max_new_tokens=50,
        do_sample=True,
        temperature=0.7
    )
t_native = (time.perf_counter() - t0) * 1000.0
text_native = tokenizer.decode(output_native[0], skip_special_tokens=True)
print(f"--- GENERACIÓN NATIVA ---\n{text_native}")
print(f"Tiempo de generación (native): {t_native:.2f} ms\n")

# 8. Importante: Limpiar los hooks al finalizar (por ejemplo, al apagar el servidor de inferencia)
steerer.remove_hooks()
print("Hooks eliminados. Modelo restaurado a su estado nativo.")
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from transformers import AutoTokenizer, AutoModelForCausalLM

# =====================================================================
# KERNEL COMPILADO (JIT / PyTorch 2.0+)
# =====================================================================
@torch.compile(mode="max-autotune", fullgraph=True)
def _compiled_steering_kernel(
    logits: torch.Tensor,
    vocab_bias: torch.Tensor,
    gamma_max: float,
    h_target: float,
    tau: float
) -> torch.Tensor:
    """
    Kernel compilado en grafo cerrado. Fusiona el cálculo de entropía de Shannon,
    la modulación sigmoidal y la inyección tensorial sobre los logits en un solo paso GPU.
    """
    last_logits = logits[:, -1, :]

    # Entropía de Shannon base 2 (0.6931471805599453 = ln(2))
    probs = F.softmax(last_logits, dim=-1)
    log_probs = F.log_softmax(last_logits, dim=-1)
    H_t = -torch.sum(probs * log_probs, dim=-1) / 0.6931471805599453

    # Control sigmoidal en tensor puro
    h_diff = (H_t - h_target) / tau
    gamma_t = gamma_max * torch.sigmoid(h_diff).unsqueeze(-1)

    # Inyección in-place/fusionada sobre los logits del último token
    steered_logits = last_logits + (gamma_t * vocab_bias)
    logits_out = logits.clone()
    logits_out[:, -1, :] = steered_logits
    return logits_out


# =====================================================================
# CLASE PRINCIPAL DE PRODUCCIÓN
# =====================================================================
class ProductionHookSteerer:
    """
    Middleware de steering para producción en PyTorch.
    - Cero sincronización CPU <-> GPU (.item() eliminados).
    - Precalculo de bias probabilístico fuera del loop autorregresivo.
    - Kernel de modulación fusionado vía torch.compile.
    """
    def __init__(
        self,
        model: nn.Module,
        tokenizer: AutoTokenizer,
        target_layer_idx: int = 14,
        gamma_max: float = 7.0,
        h_target: float = 3.8,
        tau: float = 0.5
    ):
        self.model = model
        self.tokenizer = tokenizer
        self.target_layer_idx = target_layer_idx
        self.gamma_max = gamma_max
        self.h_target = h_target
        self.tau = tau

        self.mu_channel = None
        self.std_channel = None
        self.target_energy = None

        # Tensor de bias precalculado en GPU: [1, Vocab]
        self.precomputed_vocab_bias = None
        self.is_active = False

        self.hooks = []
        self._attach_hooks()

    def _extract_residual_energy(self, hidden_state: torch.Tensor) -> torch.Tensor:
        """Descomposición R = A - M optimizada en tensores CUDA."""
        A = hidden_state.squeeze(0).to(torch.float32)
        A_pos = A - A.min() + 1e-8

        F_vec = A_pos.sum(dim=-1, keepdim=True)
        C_vec = A_pos.sum(dim=-2, keepdim=True)
        T_sum = A_pos.sum()

        M = torch.matmul(F_vec, C_vec) / T_sum
        R = A_pos - M
        return (R ** 2).mean(dim=0).unsqueeze(0)

    def _layer_hook(self, module, input, output):
        """Hook Extractor de baja sobrecarga en capa intermedia."""
        hidden_states = output[0] if isinstance(output, tuple) else output
        self.current_R_energy = self._extract_residual_energy(hidden_states)

    def _lm_head_hook(self, module, input, output):
        """Hook Modulador: Invoca el kernel compilado si el steering está activo."""
        if not self.is_active or self.precomputed_vocab_bias is None:
            return output

        return _compiled_steering_kernel(
            logits=output,
            vocab_bias=self.precomputed_vocab_bias,
            gamma_max=self.gamma_max,
            h_target=self.h_target,
            tau=self.tau
        )

    def _attach_hooks(self):
        """Registra los forward hooks en el modelo HuggingFace/PyTorch."""
        if hasattr(self.model, "model"):
            layers = self.model.model.layers
        elif hasattr(self.model, "transformer"):
            layers = self.model.transformer.h
        else:
            raise AttributeError("Arquitectura del modelo no soportada para hooks de capa.")

        h1 = layers[self.target_layer_idx].register_forward_hook(self._layer_hook)
        h2 = self.model.get_output_embeddings().register_forward_hook(self._lm_head_hook)
        self.hooks.extend([h1, h2])

    def calibrate(self, neutral_corpus: list[str]):
        """Calcula medias y desviaciones estándar del corpus neutro."""
        energies = []
        device = next(self.model.parameters()).device

        with torch.no_grad():
            for text in neutral_corpus:
                inputs = self.tokenizer(text, return_tensors="pt").to(device)
                self.model(**inputs)
                energies.append(self.current_R_energy)

        stack = torch.cat(energies, dim=0)
        self.mu_channel = stack.mean(dim=0, keepdim=True)
        self.std_channel = stack.std(dim=0, keepdim=True) + 1e-8

    def set_target(self, anchor_texts: list[str]):
        """
        Calcula la firma de energía e incrusta la proyección tensorial precalculada
        (Delta_E @ W_head.T) directamente en memoria GPU.
        """
        if self.mu_channel is None:
            raise ValueError("Se debe ejecutar calibrate() antes de fijar el target.")

        device = next(self.model.parameters()).device
        whitened_vectors = []

        with torch.no_grad():
            for text in anchor_texts:
                inputs = self.tokenizer(text, return_tensors="pt").to(device)
                self.model(**inputs)
                E_w = (self.current_R_energy - self.mu_channel) / self.std_channel
                whitened_vectors.append(E_w)

        stack = torch.cat(whitened_vectors, dim=0)
        self.target_energy = F.normalize(stack.mean(dim=0, keepdim=True), p=2, dim=1)

        # Precalculo del sesgo de vocabulario fuera del loop de generación
        lm_head = self.model.get_output_embeddings()
        W_head = lm_head.weight.detach()  # [Vocab, D]
        Delta_E = self.target_energy.to(dtype=W_head.dtype, device=W_head.device)

        # [1, D] x [D, Vocab] -> [1, Vocab]
        self.precomputed_vocab_bias = torch.matmul(Delta_E, W_head.T)
        self.is_active = True

    def disable(self):
        """Desactiva temporalmente el control sin desenganchar los hooks."""
        self.is_active = False

    def enable(self):
        """Reorganiza la activación del control."""
        if self.precomputed_vocab_bias is not None:
            self.is_active = True

    def remove_hooks(self):
        """Elimina definitivamente los hooks de PyTorch."""
        for h in self.hooks:
            h.remove()
        self.hooks.clear()
```

## Arquitectura del Hook Steerer: `NeighborhoodHookSteerer`

La clase `NeighborhoodHookSteerer` implementa un mecanismo de "steering" o direccionamiento en modelos de lenguaje grandes (LLMs) basado en la manipulación de estados ocultos y log-probabilidades durante el proceso de generación de texto. Su objetivo es influir en la salida del modelo para que se alinee con un "concepto" o "dirección" deseada, definida por un conjunto de textos ancla.

### Componentes Clave:

1.  **Inicialización (`__init__`)**:
    *   Carga un modelo pre-entrenado de Hugging Face (`AutoModelForCausalLM`) y su tokenizer asociado.
    *   Define parámetros como `target_layer_idx` (la capa del modelo donde se extrae la información para el steering), `gamma_max` (la intensidad máxima del steering), `h_target` (entropía objetivo) y `tau` (parámetro de suavizado para la función sigmoide).
    *   Inicializa variables para almacenar estadísticas de calibración (`mu_channel`, `std_channel`), la energía del objetivo (`target_energy`), y un historial de telemetría para monitorear el proceso (`telemetry_history`).
    *   Activa los hooks del modelo a través de `_attach_hooks`.

2.  **Extracción de Energía Residual (`_extract_residual_energy`)**:
    *   Esta función toma un estado oculto (`hidden_state`) de una capa del modelo y calcula una "energía residual" (`R`).
    *   Se basa en una adaptación del algoritmo de extracción de energía para identificar patrones relevantes en las activaciones del modelo.
    *   Esta energía `R` es clave para entender las representaciones internas del modelo y cómo se desvían de un estado neutral.

3.  **Hook de Capa (`_layer_hook`)**:
    *   Se adjunta a una capa específica del modelo (`target_layer_idx`).
    *   Captura los estados ocultos de esta capa y utiliza `_extract_residual_energy` para calcular `self.current_R_energy`, que representa la energía residual en ese momento de la generación.

4.  **Hook del `lm_head` (`_lm_head_hook_optimized`)**:
    *   Este es el corazón del steering. Se adjunta a la capa final del modelo (la `lm_head`) que proyecta los estados ocultos al vocabulario para obtener los logits.
    *   **Cálculo de Entropía (`H_t`)**: Mide la incertidumbre de la distribución de probabilidad del siguiente token. Un valor bajo indica alta predictibilidad, y un valor alto, más diversidad.
    *   **Cálculo de Gamma (`gamma_t`)**: Ajusta dinámicamente la fuerza del steering en función de la entropía actual (`H_t`) y la entropía objetivo (`h_target`). Utiliza una función sigmoide para suavizar la transición. Si `H_t` es mucho mayor que `h_target`, `gamma_t` tiende a `gamma_max`, aplicando más steering.
    *   **Proyección del Bias (`vocab_bias`)**: Proyecta la energía objetivo (`target_energy`) al espacio del vocabulario, creando un "bias" que se sumará a los logits del modelo. Este bias empuja la generación hacia el concepto deseado.
    *   **Modulación de Logits**: Suma el `vocab_bias` (escalado por `gamma_t`) a los logits originales del modelo, modificando las probabilidades del siguiente token.
    *   **Telemetría**: Registra `H_t` y `gamma_t` en `self.telemetry_history` para su posterior análisis.

5.  **Adjuntar Hooks (`_attach_hooks`)**:
    *   Identifica la estructura de capas del modelo (compatible con arquitecturas como Llama, Qwen, GPT-NeoX).
    *   Registra `_layer_hook` en la `target_layer_idx` y `_lm_head_hook_optimized` en el `lm_head`.

6.  **Calibración (`calibrate`)**:
    *   Toma un `neutral_corpus` (textos que representan el comportamiento "base" del modelo).
    *   Procesa estos textos para calcular `mu_channel` (media) y `std_channel` (desviación estándar) de la energía residual.
    *   Estos valores se usan para "blanquear" (normalizar) las energías residuales de los textos ancla, haciendo que el steering sea más robusto y menos dependiente de la magnitud absoluta de las activaciones.

7.  **Establecer Objetivo (`set_target`)**:
    *   Recibe una lista de `anchor_texts` (textos que ejemplifican el concepto o estilo deseado).
    *   Procesa estos textos para calcular la energía residual, la "blanquea" usando las estadísticas de calibración, y luego calcula `target_energy` (el vector normalizado que representa la dirección de steering).

8.  **Desactivar Steering (`disable_steering`)**:
    *   Simplemente establece `self.target_energy` a `None`, lo que efectivamente desactiva la modulación en el `lm_head_hook`, haciendo que el modelo genere de forma nativa.

9.  **Remover Hooks (`remove_hooks`)**:
    *   Limpia los hooks registrados, restaurando el comportamiento original del modelo. Esto es crucial para evitar efectos secundarios no deseados o fugas de memoria.

### Funcionamiento General:

El Steerer trabaja en un bucle de feedback: durante la generación de cada token, mide la "energía" interna del modelo (a través de `_layer_hook`), compara la entropía de la salida potencial con un objetivo, y ajusta los logits de la predicción del siguiente token (a través de `_lm_head_hook_optimized`) para empujar el modelo hacia el `target_energy` definido por los textos ancla, pero con una intensidad (`gamma_t`) que se adapta a la "necesidad" de steering (baja entropía = menos steering; alta entropía = más steering para guiar la generación).
